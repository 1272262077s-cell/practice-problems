```python
#to find the maximum  and minimum elements in a tuple
numbers = (25,10,45,5,60,30)
maximum = max(numbers)
minimum = min(numbers)
print("Tuple:", numbers)
print("Maximum elements:", maximum)
print("Minimum elements:", minimum)
```

```python
#python program to swap the first and last element of a tuple

# Original tuple
t = (10, 20, 30, 40, 50)

# Swap first and last elements
t = (t[-1],) + t[1:-1] + (t[0],)

print("Original tuple: (10, 20, 30, 40, 50)")
print("After swapping:", t)
```
