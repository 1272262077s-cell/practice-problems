```python
#write a program to iterate over a dictionary using for loop amd print the keys alone, values alone and both keys and values
# Create a dictionary
my_dict = {
    "1": "a",
    "2": "b",
    "3": "c",
    "4": "d"
}

# Print keys alone
print("Keys:")
for key in my_dict:
    print(key)

# Print values alone
print("\nValues:")
for value in my_dict.values():
    print(value)

# Print both keys and values
print("\nKeys and Values:")
for key, value in my_dict.items():
    print(key, value)
```

```python
#program to prepare a dictionary where the keys are numbers between 1 and 15(both included) and the values are sq of the keys
# Create an empty dictionary
squares = {}

# Add numbers from 1 to 15 and their squares
for i in range(1, 16):
    squares[i] = i * i

print(squares)
```

```python
#write a program to sum of all the valus ien a dictionary,considering the values will of int type 
# Create a dictionary
numbers = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}

# Find the sum of all values
total = 0

for value in numbers.values():
    total = total + value

print("Sum of all values:", total)
```

```python
#assign 5 elements in dictionary named my_dict. This dictionary contains student's information. Add and delete student information  from my_dict. Display student information
# Create a dictionary with 5 student details
my_dict = {
    "name": "sakshi",
    "age": 18,
    "roll_no": 5,
    "course": "Python",
    "marks": 85
}

# Display original student information
print("Original Student Information:")
print(my_dict)

# Add new student information
my_dict["college"] = "MIT College"

print("\nAfter adding information:")
print(my_dict)

# Delete student information
del my_dict["marks"]

print("\nAfter deleting information:")
print(my_dict)

# Display final student information
print("\nFinal Student Information:")
for key, value in my_dict.items():
    print(key, ":", value)
```

```python
#sort the elements of the dictionary using() sorted function
my_dict = {"name": "sakshi", "panel": "A", "rollno": 3, 
          "marks": [67,78,89,7]}

sorted_dict = dict(sorted(my_dict.items()))

print(sorted_dict)
```

```python
#create a dictionary of 5 students with marks and display the student with the highest marks

students = {
    "ishmeet": 85,
    "drishti": 92,
    "chetana": 78,
    "sakshi": 95,
    "simran": 88
}

highest_student = max(students, key=students.get)

print("Student with highest marks:", highest_student)
print("Marks:", students[highest_student])
```

```python
#count the frequency of each element in a list using a dictionary
numbers = [1, 2, 3, 2, 1, 4, 2, 3, 1]

frequency = {}
    
for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num]+1
    else:
        frequency[num] = 1

print("frequency of elements:",frequency)
```
