```python
#data_cleaning
import pandas as pd
# define a dictionary with sample data which includes some missing values
data = {
    'A': [1, 2, 3, None, 5],
    'B': [None, 2, 3, 4, 5],
    'C': [1, 2, None, None, 5]
}

df = pd.DataFrame (data)
print ("Original Data:\n", df)
print()
# use dropna() to remove rows with any missing values
df_cleaned = df.dropna()
print(df_cleaned)
```

```python
#to fill missing values
import pandas as pd
# define a dictionary with sample data which includes some missing values
data = {
    'A': [1, 2, 3, None, 5],
    'B': [None, 2, 3, 4, 5],
    'C': [1, 2, None, None, 5]
}

df = pd.DataFrame (data)
print ("Original Data:\n", df)
#filling NaN values with 0

df.fillna(0, inplace=True)
print(df.fillna)
```

```python
#use aggregate functions to fill missing values
import pandas as pd
# define a dictionary with sample data which includes some missing values
data = {
    'A': [1, 2, 3, None, 5],
    'B': [None, 2, 3, 4, 5],
    'C': [1, 2, None, None, 5]
}

df = pd.DataFrame(data)
print ("Original Data:\n", df)
#filling NaN values with the mean of each column 

df.fillna(df.mean(), inplace=True)
print(df.fillna)
```

```python
#handle duplicate values
import pandas as po
 # sample data
data = {
'A': [1, 2, 2, 3, 3, 4],
'B': [5, 6, 6, 7, 8, 8]}
df = pd. DataFrame(data)
print( "Original DataFrame: \n", df. to_string(index=False))
# detect duplicates
print("Duplicate Rows: In", df[df.duplicated()]. to_string(index
=False))
# remove duplicates based on column 'A'
df.drop_duplicates(subset=['A'], keep='first', inplace=True) 
print ("DataFrame after removing duplicates based on column 'A':\n",
       df.to_string(index=False))
```

```python
#rename column names to meaningful names
import pandas as pd
# sample data
data = {
    'A': [25, 30, 35],
    'B': ['Simran','Drishti','Chetana'],
    'C': [50000, 60000, 70000]
}
df = pd.DataFrame(data)
# rename columns
df.rename(columns={'A': 'Age', 'B':'Name', 'C': 'Salary'},
   inplace=True)

print(df.to_string(index=False))
```
