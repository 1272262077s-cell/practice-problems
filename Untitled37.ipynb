```python
#3 accept the no. from user and display prime factors of that no.
n = int(input("Enter a number: "))

def prime_factors(n):
    factors = []
    i = 2

    while i <= n:
        if n % i == 0:
            factors.append(i)
            n //= i
        else:
            i += 1

    return factors

print("Prime factors:", prime_factors(n))
```

```python
#recursion
def print_numbers(n):
    if n==0:     #base condition
        return
    print_numbers(n-1) #recursive call
    print(n)
n=int(input("enter n:"))
print_numbers(n)
```

```python
#recursion in factorial
def factorial(n):
    if n==0 or n==1:
        return 1
    return n * factorial(n-1) 
n=int(input("enter a number:"))
print("Factorial=",factorial(n))
```

```python
def fibonacci (n):
    if n <= 1:
         return n
    return fibonacci(n- 1) + fibonacci(n - 2)
n = int(input("Enter number of terms: "))

for i in range(n):
    print(fibonacci(i), end="")
```
